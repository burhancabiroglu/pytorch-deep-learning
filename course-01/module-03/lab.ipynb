```python
from torch.utils.data import DataLoader, Dataset, random_split
import os
import scipy
from PIL import Image
from torchvision import transforms
```

```python
class OxfordFlowerDataset(Dataset):
    def __init__(self, root_dir: str):
        self.root_dir = root_dir
        self.img_dir = os.path.join(root_dir, 'jpg')

        labels_mat = scipy.io.loadmat(os.path.join(root_dir,'imagelabels.mat'))

        self.labels = labels_mat['labels'][0] - 1

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        image_name = f'image_{idx:05d}.jpg'
        image_path = os.path.join(self.img_dir, image_name)

        # load image
        image = Image.open(image_path)
        label = self.labels[idx]
        return image, label
```

```python

transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    # transforms.Normalize(...)
])

dataset = OxfordFlowerDataset('./flower_data', transform = transform)
```

```python
# 0.7 0.15, kalan

dataset_len = len(dataset)
train_size = int(dataset_len * 0.7)
val_size = int(dataset_len * 0.15)
test_size = int(dataset_len * 0.15)
train_dataset, val_dataset, test_dataset = random_split(dataset, [train_size, val_size ,test_size])

loader = DataLoader(dataset=dataset, batch_size= 32, shuffle=True)
```
